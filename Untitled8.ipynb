```python
import time
import copy
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, random_split
import torchvision
from torchvision import transforms
import matplotlib.pyplot as plt

torch.manual_seed(42)
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Urządzenie:", device)

mean = (0.4914, 0.4822, 0.4465)
std  = (0.2470, 0.2435, 0.2616)

tf = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

full_train_ds = torchvision.datasets.CIFAR10("./data", train=True, download=False, transform=tf)
test_ds       = torchvision.datasets.CIFAR10("./data", train=False, download=False, transform=tf)


train_ds, val_ds = random_split(full_train_ds, [45000, 5000], generator=torch.Generator().manual_seed(42))

train_dl = DataLoader(train_ds, batch_size=128, shuffle=True,  num_workers=2, pin_memory=True)
val_dl   = DataLoader(val_ds,   batch_size=256, shuffle=False, num_workers=2, pin_memory=True)
test_dl  = DataLoader(test_ds,  batch_size=256, shuffle=False, num_workers=2, pin_memory=True)


class BaselineMLP(nn.Module):
    def __init__(self, in_dim=3*32*32, n_classes=10):
        super().__init__()
        self.net = nn.Sequential(
            nn.Flatten(),
            nn.Linear(in_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Linear(256, n_classes)
        )
    def forward(self, x):
        return self.net(x)

class CustomCNN(nn.Module):
    def __init__(self, n_classes=10):
        super().__init__()
        self.features = nn.Sequential(
            # Blok 1: 32x32 -> 16x16
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),
            # Blok 2: 16x16 -> 8x8
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2),
            # Blok 3: 8x8 -> 4x4
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, n_classes)
        )
    def forward(self, x):
        return self.classifier(self.features(x))

def count_params(model):
    return sum(p.numel() for p in model.parameters())

def evaluate(model, dl):
    model.eval()
    correct = total = 0
    with torch.no_grad():
        for xb, yb in dl:
            xb, yb = xb.to(device), yb.to(device)
            preds = model(xb).argmax(dim=1)
            correct += (preds == yb).sum().item()
            total += yb.numel()
    return correct / total

def train_and_select(model, epochs=12, lr=1e-3):
    model = model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    val_accs = []
    best_val_acc = 0.0
    best_weights = copy.deepcopy(model.state_dict())
    best_epoch = 1

    t0 = time.time()
    for epoch in range(1, epochs + 1):
        model.train()
        for xb, yb in train_dl:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            optimizer.step()

        val_acc = evaluate(model, val_dl)
        val_accs.append(val_acc)

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_weights = copy.deepcopy(model.state_dict())
            best_epoch = epoch

        print(f"Epoka {epoch:2d} | Val Accuracy: {val_acc*100:.2f}%")

    train_time = time.time() - t0

    model.load_state_dict(best_weights)
    test_acc = evaluate(model, test_dl)
    return val_accs, best_epoch, test_acc, train_time

EPOCHS = 12
print("=== Trening MLP (Baseline) ===")
mlp = BaselineMLP()
val_mlp, ep_mlp, test_mlp, time_mlp = train_and_select(mlp, epochs=EPOCHS)

print("\n=== Trening CustomCNN (3 bloki splotowe) ===")
cnn = CustomCNN()
val_cnn, ep_cnn, test_cnn, time_cnn = train_and_select(cnn, epochs=EPOCHS)


print(f"{'Model':12} | {'Parametry':>12} | {'Najl. Epoka':>11} | {'Test Acc':>9} | {'Czas [s]':>9}")
print("-" * 64)
print(f"{'MLP':12} | {count_params(mlp):>12,} | {ep_mlp:>11d} | {test_mlp*100:>8.2f}% | {time_mlp:>9.1f}")
print(f"{'CustomCNN':12} | {count_params(cnn):>12,} | {ep_cnn:>11d} | {test_cnn*100:>8.2f}% | {time_cnn:>9.1f}")


plt.figure(figsize=(8, 4.5))
plt.plot(range(1, EPOCHS+1), [a*100 for a in val_mlp], "o-", label=f"MLP (Test: {test_mlp*100:.2f}%)")
plt.plot(range(1, EPOCHS+1), [a*100 for a in val_cnn], "s-", label=f"CustomCNN (Test: {test_cnn*100:.2f}%)")
plt.axhline(10, color="gray", linestyle="--", linewidth=1, label="Baseline losowy (10%)")
plt.xlabel("Epoka")
plt.ylabel("Accuracy walidacyjna [%]")
plt.title("Porównanie MLP vs własny CustomCNN na zbiorze CIFAR-10")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
